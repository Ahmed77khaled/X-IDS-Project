<p style="margin-bottom: 8px;">طب إيه المشكلة لو سبنا القيم دي موجودة في البيانات؟ المشكلة إن بعض أدوات معالجة البيانات وخوارزميات تعلم الآلة مش بتقدر تتعامل مع القيم غير المحدودة، ووجودها ممكن يسبب أخطاء أثناء تجهيز البيانات أو تدريب النموذج (زي خطأ <bdi style="font-family: 'Cascadia Code', monospace; color: #f85149;">Overflow</bdi>). وعشان كده، قبل ما نكمل، لازم نعالج المشكلة دي.</p>

<div style="color: #c9d1d9; font-size: 0.92em; line-height: 1.85; user-select: text; -webkit-user-select: text; cursor: text;">
<p style="margin-bottom: 8px;">تمام يا هندسة، بعد ما جرّبنا حذف الصفوف اللي فيها قيم مفقودة في الخطوة اللي فاتت، وشوفنا إن الطريقة دي ممكن تخلّينا نخسر سجلات مهمة من الداتاسيت، جه وقت نستخدم حل أذكى، نحافظ بيه على البيانات ونعالج القيم الناقصة في نفس الوقت.</p>
<p style="margin-bottom: 8px;">في الخطوة التامنة، هنتعرف على طريقة إحصائية مهمة اسمها التعويض بالوسيط، أو <strong>ميديان (<bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">Median</bdi>)</strong>، وهنستخدمها عشان نملأ القيم المفقودة الموجودة في الخصائص.</p>
<p style="margin-bottom: 8px;">طب ليه هنستخدم الوسيط تحديداً؟ وليه مش المتوسط الحسابي، أو مين (<bdi style="font-family: 'Cascadia Code', monospace; color: #f85149;">Mean</bdi>)؟ خليني أوضحلك الفرق: المتوسط الحسابي بيتحسب بجمع كل القيم وقسمتها على عددها، وده معناه إنه بيتأثر جداً بوجود قيم كبيرة أو صغيرة بشكل غير معتاد. وده خطير جداً في بيانات الشبكات، لأن حركة المرور بتحتوي على قيم متطرفة شاذة (<bdi style="font-family: 'Cascadia Code', monospace; color: #f85149;">Outliers</bdi>) نتيجة هجمات حجب الخدمة (<bdi style="font-family: 'Cascadia Code', monospace; color: #f85149;">DoS</bdi>)، فالمتوسط هيتشوه ويقفز لأرقام غير طبيعية. أما الوسيط فهو إحصاء متين (<bdi style="font-family: 'Cascadia Code', monospace; color: #3fb950;">Robust Statistic</bdi>) لأنه بيمثل القيمة اللي في المنتصف ولا يتأثر بالانفجارات اللحظية.</p>
<p style="margin-bottom: 8px;">نيجي بقى للتطبيق العملي: بما إننا فصلنا الخصائص والتصنيفات في الخطوة الخامسة، فإحنا هنشتغل هنا على مصفوفة <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">X</bdi>، اللي بتحتوي على خصائص حركة الشبكة.</p>
<p style="margin-bottom: 8px;">• أول حاجة، هنحسب وسيط الأعمدة الرقمية: <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">median_values = X.median(numeric_only=True)</bdi>.</p>
<p style="margin-bottom: 8px;">• بعد كده، هنستخدم دالة <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">fillna()</bdi> لتعويض القيم الناقصة: <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">X = X.fillna(median_values)</bdi>. وبكده أي خانة فيها <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">NaN</bdi> هتتعوض بالوسيط الخاص بعمودها.</p>
<p style="margin-bottom: 8px;">لكن هل كده المشكلة اتحلت خلاص؟ مش قبل ما نتأكد! وعشان كده، هنستخدم <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">X.isna().sum().sum()</bdi> لحساب إجمالي عدد القيم المفقودة في مصفوفة <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">X</bdi> بالكامل، والمفروض النتيجة تطلع <strong>صفر (0)</strong> تماماً!</p>
<p style="margin-bottom: 8px;">دلوقتي دورك يا هندسة: احسب وسيط الأعمدة الرقمية، وعوض القيم بـ <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">fillna()</bdi>، واطبع إجمالي القيم المفقودة بـ <bdi style="font-family: 'Cascadia Code', monospace; color: #58a6ff;">isna().sum().sum()</bdi>، واضغط <bdi style="font-family: 'Cascadia Code', monospace; background: #27272a; color: #fff; padding: 2px 7px; border-radius: 4px; border: 1px solid #3f3f46;">Shift + Enter</bdi>.</p>
<p style="margin-bottom: 8px;">زي الفل، كده إحنا عالجنا مشكلة القيم المفقودة وبقينا أقرب لتجهيز الداتاسيت بشكل نهائي. تفتكر إيه اللي ممكن يحصل لو النموذج حفظ عناوين الآي بي وأرقام المنافذ بدل ما يفهم أنماط الهجوم نفسها؟ دي نقطة مهمة، وهنفك شفرتها في الخطوة التاسعة!</p>

```python
# 1. Calculate numeric medians and impute missing values
numeric_medians = df.median(numeric_only=True)
df.fillna(numeric_medians, inplace=True)
# 2. Verify all missing values have been eliminated (Expected: 0)
remaining_nulls = df.isna().sum().sum()
print("Remaining missing values:", remaining_nulls)
```
